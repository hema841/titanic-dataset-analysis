```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Hema', 'Ravi', 'Priya', 'Kiran'],
    'Age': [20, 21, 19, 22]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Marks': [85, 90, 88, 76],
    'City': ['Vizag', 'Hyderabad', 'Chennai', 'Delhi']
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python
inner_merge = pd.merge(df1, df2, on='ID', how='inner')
print(inner_merge)
```

```python
left_merge = pd.merge(df1, df2, on='ID', how='left')
print(left_merge)
```

```python
right_merge = pd.merge(df1, df2, on='ID', how='right')
print(right_merge)
```

```python
outer_merge = pd.merge(df1, df2, on='ID', how='outer')
print(outer_merge)
```

```python
df3 = pd.DataFrame({
    'ID': [6, 7],
    'Name': ['Anu', 'Rohit'],
    'Age': [20, 23]
})

vertical_concat = pd.concat([df1, df3], axis=0, ignore_index=True)
print(vertical_concat)
```

```python
df4 = pd.DataFrame({
    'Grade': ['A', 'A+', 'B+', 'A']
})
horizontal_concat = pd.concat([df1, df4], axis=1)
print(horizontal_concat)
```

```python
students = pd.DataFrame({
    'Name': ['Hema', 'Ravi', 'Priya']
}, index=[1, 2, 3])

marks = pd.DataFrame({
    'Marks': [85, 90, 88]
}, index=[1, 2, 3])
result = students.join(marks)
print(result)
```

```python
filled_data = outer_merge.copy()
filled_data['Name'] = filled_data['Name'].fillna('Unknown')
filled_data['Age'] = filled_data['Age'].fillna(
    filled_data['Age'].mean()
)
filled_data['Marks'] = filled_data['Marks'].fillna(0)
filled_data['City'] = filled_data['City'].fillna('Not Available')
print(filled_data)
```

```python
duplicate_df = pd.DataFrame({
    'ID': [1, 2, 2, 3],
    'Name': ['Hema', 'Ravi', 'Ravi', 'Priya']
})
print("Original DataFrame:")
print(duplicate_df)
cleaned_df = duplicate_df.drop_duplicates()
print("\nAfter removing duplicates:")
print(cleaned_df)
```
