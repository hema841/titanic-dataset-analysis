```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
df['species'] = pd.Categorical.from_codes(
    iris.target, iris.target_names
)
print(df.head())
print("\nDataset shape:", df.shape)
print("\nSpecies counts:")
print(df['species'].value_counts())
sns.set_theme(style="whitegrid")
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species',
    s=80
)

plt.title('Iris Dataset: Sepal Measurements')
plt.show()
```

```python

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x=df.index,
    y='sepal length (cm)',
    hue='species'
)

plt.title('Sepal Length Across Observations')
plt.xlabel('Observation Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='petal length (cm)',
    hue='species',
    bins=15,
    kde=True,
    element='step'
)

plt.title('Distribution of Petal Length')
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='species',
    y='petal length (cm)'
)

plt.title('Petal Length by Species')
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='species',
    y='sepal width (cm)',
    hue='species',
    legend=False
)

plt.title('Sepal Width Distribution by Species')
plt.show()
```

```python

sns.pairplot(
    df,
    hue='species',
    diag_kind='hist'
)

plt.show()
```

```python

plt.figure(figsize=(8, 6))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Iris Feature Correlation Heatmap')
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='species',
    y='petal length (cm)',
    errorbar=None
)

plt.title('Average Petal Length by Species')
plt.ylabel('Average Petal Length (cm)')
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x='species',
    hue='species',
    legend=False
)

plt.title('Number of Flowers per Species')
plt.ylabel('Count')
plt.show()
```

```python

sns.jointplot(
    data=df,
    x='petal length (cm)',
    y='petal width (cm)',
    hue='species',
    height=7
)

plt.show()
```

```python

features = iris.feature_names

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, feature in zip(axes.flat, features):
    sns.histplot(
        data=df,
        x=feature,
        hue='species',
        element='step',
        ax=ax
    )
    ax.set_title(feature)

plt.tight_layout()
plt.show()
```
