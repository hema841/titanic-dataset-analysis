```python
import pandas as pd
df1 = pd.DataFrame({
    'Name': ['Anu', 'Ravi'],
    'Marks': [85, 78]
})

df2 = pd.DataFrame({
    'Name': ['Priya', 'Kiran'],
    'Marks': [92, 88]
})
# Combine rows
result = pd.concat([df1, df2], ignore_index=True)
print(result)
```

```python
result = pd.concat([df1, df2], axis=1)
print(result)
```

```python
students = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Anu', 'Ravi', 'Priya']
})
marks = pd.DataFrame({
    'ID': [1, 2, 4],
    'Marks': [85, 78, 90]
})
result = pd.merge(students, marks, on='ID', how='inner')
print(result)
```

```python
# Keep all students, including those without marks
left_result = pd.merge(
    students, marks, on='ID', how='left'
)
# Keep every ID from both DataFrames
outer_result = pd.merge(
    students, marks, on='ID', how='outer'
)
```

```python
df1 = pd.DataFrame(
    {'Name': ['Anu', 'Ravi']},
    index=[1, 2]
)
df2 = pd.DataFrame(
    {'Marks': [85, 78]},
    index=[1, 2]
)
result = df1.join(df2)
print(result)
```

```python
df1 = pd.DataFrame({
    'A': [1, None],
    'B': [None, 4]
})
df2 = pd.DataFrame({
    'A': [10, 20],
    'B': [30, 40]
})
result = df1.combine_first(df2)
print(result)
```

```python
df = pd.DataFrame({
    'Name': ['Anu', 'Anu', 'Ravi', 'Ravi'],
    'Subject': ['Maths', 'Science', 'Maths', 'Science'],
    'Marks': [85, 90, 78, 82]
})

result = df.pivot(
    index='Name',
    columns='Subject',
    values='Marks'
)
print(result)
```

```python
df = pd.DataFrame({
    'Department': ['CSE', 'CSE', 'ECE', 'ECE'],
    'Gender': ['M', 'F', 'M', 'F'],
    'Marks': [80, 90, 70, 85]
})
result = pd.pivot_table(
    df,
    values='Marks',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)
print(result)
```

```python
df = pd.DataFrame({
    'Name': ['Anu', 'Ravi'],
    'Maths': [85, 78],
    'Science': [90, 82]
})

result = pd.melt(
    df,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'
)

print(result)
```

```python

import pandas as pd
df = pd.DataFrame({
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'CSE', 'ECE'],
    'Marks': [85, 78, 92, 88]
})
df = df.fillna(0)
df = df.drop_duplicates()
df = df.rename(columns={'Name': 'Student_Name'})
print("Sorted DataFrame:")
print(df.sort_values(by='Marks', ascending=False))
print("\nStudents scoring 80 or more:")
print(df[df['Marks'] >= 80])
print("\nAverage marks by department:")
print(df.groupby('Department')['Marks'].mean())
df.to_csv('processed_data.csv', index=False)
print("\nData saved successfully!")
```
